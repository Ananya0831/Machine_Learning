```python
CSV_PATH = "/content/drive/MyDrive/SEM - 4/ML - datasets/titanic.csv"

df = pd.read_csv(CSV_PATH)

df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
fig1 = plt.figure(figsize=(10,6))

sns.heatmap(
    df.isnull(),
    cmap="viridis",
    cbar=False,
    yticklabels=False
)

plt.title("Missing Values Heatmap - Titanic Dataset")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
fig2, axes = plt.subplots(2,3,figsize=(15,8))

cols = [
    'age',
    'fare',
    'sibsp',
    'parch',
    'pclass',
    'survived'
]

for i,col in enumerate(cols):

    r,c = divmod(i,3)

    sns.histplot(
        df[col].dropna(),
        bins=25,
        kde=True,
        ax=axes[r,c]
    )

    axes[r,c].set_title(col.title())

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
fig3, axes = plt.subplots(2,2,figsize=(12,8))

sns.countplot(x='sex',data=df,ax=axes[0,0])
axes[0,0].set_title("Gender Distribution")

sns.countplot(x='pclass',data=df,ax=axes[0,1])
axes[0,1].set_title("Passenger Class")

sns.countplot(x='survived',data=df,ax=axes[1,0])
axes[1,0].set_title("Survival Distribution")

sns.countplot(
    x='sex',
    hue='survived',
    data=df,
    ax=axes[1,1]
)

axes[1,1].set_title("Gender vs Survival")

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
fig4 = plt.figure(figsize=(10,8))

numeric_df = df.select_dtypes(include=['number'])

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    square=True
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
fig5, axes = plt.subplots(2,2,figsize=(12,8))

sns.boxplot(y=df['age'],ax=axes[0,0])
axes[0,0].set_title("Age")

sns.boxplot(y=df['fare'],ax=axes[0,1])
axes[0,1].set_title("Fare")

sns.boxplot(y=df['sibsp'],ax=axes[1,0])
axes[1,0].set_title("SibSp")

sns.boxplot(y=df['parch'],ax=axes[1,1])
axes[1,1].set_title("Parch")

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
fig6 = plt.figure(figsize=(8,6))

sns.scatterplot(
    x='age',
    y='fare',
    hue='survived',
    data=df
)

plt.title("Age vs Fare")

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
fig7 = plt.figure(figsize=(6,6))

survival = df['survived'].value_counts()

plt.pie(
    survival,
    labels=["Did Not Survive","Survived"],
    autopct="%1.1f%%",
    startangle=90,
    explode=(0.05,0)
)

plt.title("Survival Percentage")

plt.show()
```
