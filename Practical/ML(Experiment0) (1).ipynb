```python
# -- ARITHMETIC OPERATIONS --
a=7
b=2
print("a+b=",a+b)
print("a-b=",a-b)
print("a*b=",a*b)
print("a/b=",a/b)
print("a**2=",a**2)

# -- COMPARISION OPERATIONS --
cgpa=8.4
print("cgpa == 8.4:",cgpa==8.4)
print("cgpa > 8.4:",cgpa>8)
print("cgpa < 8.4:",cgpa<6)

# -- LOGICAL OPERATORS : and / or / not --
attendance=85
placed=(cgpa>8) and (attendance>80)  #both conditions must be true
print("placed: ",placed)
```
